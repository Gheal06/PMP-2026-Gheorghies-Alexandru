Să presupunem că investigăm o boală (B), care afectează doar un procent din populaţie (P(B) = 0.01). Avem un test de
diagnosticare pentru această boală, care are următoarele caracteristici:

- Sensibilitate (probabilitatea ca testul să fie pozitiv dacă persoana este bolnavă): P(Test = Pozitiv∣B) = 0.95
- Specificitate (probabilitatea ca testul să fie negativ dacă persoana nu este bolnavă): P(Test = Negativ∣¬B) = 0.90

a) Dacă o persoană este testată şi rezultatul este pozitiv, care este probabilitatea ca aceasta să fie efectiv bolnavă?
Cum explicaţí rezultatul?

b) Modelaţi în Python probabilitatea de mai sus şi comparaţi cu cea obţinută teoretic la punctul anterior.

c) Care ar trebui să fie specificitatea minimă pentru ca probabilitatea de mai sus să ajungă la 50%?

Considerăm următoarele evenimente binare:
- $T$: Testul este pozitiv
- $B$: Pacientul este bolnav

$P(T|B)=\frac{P(T,B)}{P(B)} \Leftrightarrow 0.95 = \frac{P(T,B)}{0.01} \Rightarrow P(T,B)=0.0095$

$P(\lnot T|B)=\frac{P(\lnot T,B)}{P(B)} \Leftrightarrow 0.05 = \frac{P(\lnot T,B)}{0.01} \Rightarrow P(\lnot T,B)=0.0005$

$P(T| \lnot B)=\frac{P(T, \lnot B)}{P(\lnot B)} \Leftrightarrow 0.10 = \frac{P(T, \lnot B)}{0.99} \Rightarrow P(T, \lnot B)=0.099$

$P(\lnot T| \lnot B)=\frac{P(\lnot T,\lnot B)}{P(\lnot B)} \Leftrightarrow 0.90 = \frac{P(\lnot T, \lnot B)}{0.99} \Rightarrow P( \lnot T, \lnot B)=0.891$

$P(T)=P(T,B)+P(T,\lnot B)=0.0095+0.099=0.1085$

a) $P(B|T)=\frac{P(T,B)}{P(T)}=\frac{0.0095}{0.1085} \approx 0.0875576$

Acest rezultat poate fi explicat de rata foarte mica de incidență a bolii. Intuitiv, dintr-o populație 99% sănătoasă, aprox. 5% din populație vor testa pozitiv, comparativ cu 1% din populație care chiar este bolnavă. Acest lucru ar sugera că probabilitatea ca o persoană să fie bolnavă daca a testat pozitiv este cel mult aprox. 16%.


```python
import numpy as np
def run_experiment(population: int, sick: int, sensibility: float, specificity: float):
    correct_positive = 0
    false_positive = 0
    false_negative = 0
    correct_negative = 0
    for i in range(population):
        if i < sick: # bolnav
            if np.random.rand()<sensibility:
                correct_positive += 1
            else:
                false_negative += 1
        else:
            if np.random.rand()<specificity:
                correct_negative += 1
            else:
                false_positive += 1
    
    return [[correct_positive, false_negative], [false_positive, correct_negative]]

print(run_experiment(10000000,100000,0.95,0.90))
```

$P(T)=P(T,B)+P(T,\lnot B)=0.0095+0.99-d$

$0.5 \le P(B|T)=\frac{P(T,B)}{P(T)}=\frac{0.0095}{0.0095+0.99-d}$

Deci:

$$0.5 \le \frac{0.0095}{0.0095+0.99-P(\lnot T | \lnot B) \cdot 0.99} \Leftrightarrow$$

$$\Leftrightarrow 0.0095+0.99-P(\lnot T | \lnot B)\cdot 0.99 \le 0.0190 \Leftrightarrow$$

$$\Leftrightarrow 0.0095+0.99-0.0190 \le P(\lnot T | \lnot B)\cdot 0.99 \Leftrightarrow$$
$$\Leftrightarrow \frac{0.9805}{0.99} \le P(\lnot T | \lnot B) \Leftrightarrow$$
