f) Abordarea greedy este evident greșită. De exemplu, pentru următorul exemplu:

$\pi=[0.5, 0.5], a=\begin{bmatrix} 0 & 1 \\ 1 & 0\end{bmatrix}, b=\begin{bmatrix} 0.4 & 0.6 \\ 0.3 & 0.7\end{bmatrix}, o=[1, 1, 0, 1, 0, 1, 0]$

Singurele liste de stări ascunse sunt $[1,0,1,0,1,0,1]$ și $[0,1,0,1,0,1,0]$, ambele putând fi alese cu probabilitate $\frac{1}{2}$.

Algoritmul greedy va alege prima stare ascunsă ca fiind $1$, deoarece $0.35 = \pi_1 \cdot b_{11} > \pi_0 \cdot b_{01}=0.3$.

Probabilitatea ca lista de stări ascunse $[1,0,1,0,1,0,1]$ să genereze secvența observată $o=[1, 1, 0, 1, 0, 1, 0]$ este $0.7 \cdot 0.3^3 \cdot 0.6^3 \approx 0.0040824$.

Probabilitatea ca lista de stări ascunse $[0,1,0,1,0,1,0]$ să genereze secvența observată $o=[1, 1, 0, 1, 0, 1, 0]$ este $0.6 \cdot 0.4^3 \cdot 0.7^3 \approx 0.0131712$.

Alegând greedy prima stare ascunsă vom obține o probabilitate de aproximativ $3$ ori mai mică decât optimul global. Pot fi construite ușor și exemple unde alegerea greedy la o stare intermediară este neoptimă.

Algoritmul Viterbi vrea să maximizeze $P(x_1, x_2, \ldots, x_T | o_1, o_2, \ldots, o_T)\stackrel{\text{prob. cond.}}{=}\frac{P(x_1, x_2, \ldots, x_T, o_1, o_2, \ldots, o_T)}{P(o_1, o_2, \ldots, o_T)} \propto P(x_1, x_2, \ldots, x_T, o_1, o_2, \ldots, o_T)$. Acest artificiu funcționează deoarece numitorul $P(o_1, o_2, \ldots, o_T)$ este identic la toate probabilitățile condiționate care pot fi obținute în acest fel.
