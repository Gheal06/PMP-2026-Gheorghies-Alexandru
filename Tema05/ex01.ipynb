Exercitiul 1. Cum calculăm probabilitatea unei secvente de observatii? (Forward)

Considerăm un model Markov ascuns cu stări ascunse $X_1, \ldots , X_T$ si observații $O_1, \ldots ,O_T$. Notăm prin $\pi_i = P (X_1 =
i)$ distribuția inițială, prin $a_{ij} = P (X_{t+1} = j | X_{t = i})$ probabilitățile de tranziție si prin $b_j (o) = P (O_t = o | X_t = j)$ probabilitățile de emisie.

Pentru o secvență de observații $o_1, \ldots , o_T$ , definim variabila Forward $\alpha_t (j) = P (O_1 = o_1, \ldots ,O_t = o_t, X_t = j)$.

a) Inițializare. Explicați de ce $\alpha_1(j)=\pi_jb_j (o_1)$.

b) Recurență. Folosind formula probabilității totale si proprietatea Markov, demonstrați că

$$\alpha_t (j) = b_j (o_t)\sum_{i}\alpha_{t−1} (i) a_{ij}, t = 2,\ldots,T$$

Indiciu: conditionați după starea ascunsă $X_{t−1}$ si separați probabilitatea observației curente.

c) Finalizare. Arătați că probabilitatea întregii secvențe observate este
$$P(O_1 = o_1,\ldots,O_T = o_T)= \sum_j \alpha_T (j).$$

a) Inițializare. Explicați de ce $\alpha_1(j)=\pi_jb_j (o_1)$.

#### Rezolvare

$$\alpha_1 (j) \stackrel{def.}{=} P (O_1 = o_1,X_1 = j)\stackrel{\text{prob. cond.}}{=}P(X_1=j) \cdot P(O_1=o_1|X_1=j)\stackrel{def.}{=}\pi_j \cdot b_j(o_1)$$


b) Recurență. Folosind formula probabilității totale si proprietatea Markov, demonstrați că

$$\alpha_t (j) = b_j (o_t)\sum_{i}\alpha_{t−1} (i) a_{ij}, t = 2,\ldots,T$$

#### Rezolvare

Pentru orice $t \ge 2$:
$$\alpha_t (j)\stackrel{def.}{=}P(O_1 = o_1, \ldots ,O_t = o_t, X_t = j)\stackrel{\text{prob. cond.}}{=}P(O_t=o_t|O_1 = o_1, \ldots ,O_{t-1} = o_{t-1}, X_t=j) \cdot P(O_1 = o_1, \ldots ,O_{t-1} = o_{t-1}, X_t=j)$$
$$\stackrel{\text{prop. Markov, prob. marg.}}{=}P(O_t = o_t|X_t=j) \cdot \sum_{i} P(O_1 = o_1, \ldots ,O_{t-1} = o_{t-1},X_{t-1}=i,X_t=j)$$
$$\stackrel{\text{prob. cond, def.}}{=}b_j(o_t) \cdot \sum_{i} P(X_t=j | O_1 = o_1, \ldots ,O_{t-1} = o_{t-1},X_{t-1}=i) \cdot P(O_1 = o_1, \ldots ,O_{t-1} = o_{t-1},X_{t-1}=i)$$
$$\stackrel{\text{prop. Markov, def.}}{=}b_j(o_t) \cdot \sum_{i} P(X_t=j | X_{t-1}=i) \cdot \alpha_{t-1}(i)\stackrel{\text{def.}}{=}b_j(o_t) \cdot \sum_{i} \alpha_{t-1}(i) \cdot \alpha_{ij}$$


c) Finalizare. Arătați că probabilitatea întregii secvențe observate este
$$P(O_1 = o_1,\ldots,O_T = o_T)= \sum_j \alpha_T (j)$$

#### Rezolvare

$$P(O_1 = o_1,\ldots,O_T = o_T) \stackrel{\text{prob. marg.}}=\sum_j P(O_1 = o_1,\ldots,O_T = o_T, X_T=j)\stackrel{def.}{=}\sum_j \alpha_T (j)$$
