```python
#a)
import numpy as np
def run():
    turn = np.random.randint(0,1)
    p1 = np.random.randint(1,7)
    p2 = 0
    for _ in range(p1*2):
        if turn == 0: # moneda masluita
            p2 += np.random.rand() < 4/7
        else:
            p2 += np.random.randint(0, 2)
    if turn:
        p1, p2 = p2, p1
    return 0 if p1 >= p2 else 1

def runs(iter_cnt):
    wins = [0, 0]
    for _ in range(iter_cnt):
        wins[run()] += 1
    print(wins)

runs(10000)
```
