```python
# adaptat din ex. 2 din lab 4
import math
import numpy as np
clici_a = [[0,1], [1,2], [2,3], [3,4]] # A-B-C-D-E
clici_b = [[0,1,2], [1,3], [2,4]] # A-B-C; A-C; B-D; C-E
def phi(clici, msk, beta):
    s = 0
    for i in clici:
        for j in clici:
            if i<j:
                s += ((msk>>i)&1)*((msk>>j)&1) # fiecare variabila poate fi 0 sau 1
    return math.exp(s*beta)

def prob_comuna(beta, clici):
    Z = 0
    for msk in range(32):
        prod = 1
        for c in clici:
            prod *= phi(c, msk, beta)
        Z += prod
    p = []
    for msk in range(32):
        prod = 1
        for c in clici:
            prod *= phi(c, msk, beta)
        p.append(prod/Z)
    print(p)
    print(np.sum(np.array(p)))
    return p

def infer(p, variable, evidence):
    numitor = 0
    evidence_and_variable = (1<<variable) | evidence
    for msk in range(32):
        if (msk&evidence) == evidence:
            numitor += p[msk]

    numarator = 0
    for msk in range(32):
        if (msk&evidence_and_variable) == evidence_and_variable:
            numarator += p[msk]

    return numarator/numitor

# consider ca P(A|B,C) = P(A=1|B=1, C=1) = P(A=1,B=1,C=1)/P(B=1,C=1) = suma din P(A=1, B=1, C=1, D=d, E=e)/suma din P(A=a, B=1, C=1, D=d, E=e)
# rationamentul este similar pentru P(D|A,B,C,E)

print(f"P(A|B,C)={infer(prob_comuna(beta=1, clici=clici_a), 0, (1<<1) | (1<<2))}") # P(A|B,C)
print(f"P(D|A,B,C,E)={infer(prob_comuna(beta=1, clici=clici_b), 3, 31-(1<<3))}") # P(D|A,B,C,E)
```
