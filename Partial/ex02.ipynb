```python
# adaptat din ex. 2 din lab 4
import math
import numpy as np
clici_a = [[0,1], [1,2], [2,3], [3,4]] # A-B-C-D-E
clici_b = [[0,1,2], [1,3], [2,4]] # A-B-C; A-C; B-D; C-E
def phi(clici, msk, beta):
    s = 0
    for i in clici:
        for j in clici:
            if i<j:
                s += ((msk>>i)&1)*((msk>>j)&1) # fiecare variabila poate fi 0 sau 1
    return math.exp(s*beta)

def prob_comuna(beta, clici):
    Z = 0
    for msk in range(32):
        prod = 1
        for c in clici:
            prod *= phi(c, msk, beta)
        Z += prod
    p = []
    for msk in range(32):
        prod = 1
        for c in clici:
            prod *= phi(c, msk, beta)
        p.append(prod/Z)
    print(p)
    print(np.sum(np.array(p)))
    return p

def infer(p, variable, evidence):
    numitor = [0]*32
    for msk in range(32):
        numitor[msk&evidence] += p[msk]

    numarator = [0]*32
    for msk in range(32):
        if msk >> variable & 1:
            numarator[msk&evidence] += p[msk]

    s = 0
    for msk in range(32):
        if msk & evidence == msk: 
            s += numarator[msk]/numitor[msk]
    return s


print(f"P(D|B,C)={infer(prob_comuna(beta=1, clici=clici_a), 0, (1<<1) | (1<<2))}") # P(A|B,C)
print(f"P(D|A,B,C,E)={infer(prob_comuna(beta=1, clici=clici_b), 3, 31-(1<<3))}") # P(D|A,B,C,E)
```
