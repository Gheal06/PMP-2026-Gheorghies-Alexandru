```python
# adaptare a codului de la ex 2 din tema 5
import numpy as np
initial_prob = np.asarray([1/4, 1/4, 1/4, 1/4], dtype=np.float64)
transition = np.asarray([
    [0.2,0.4,0.0,0.4],
    [0.4,0.2,0.4,0.0],
    [0.0,0.4,0.2,0.4],
    [0.4,0.0,0.4,0.2],
], dtype=np.float64)
# 0 = red
# 1 = green
# 2 = blue
emission_prob = np.asarray([
    [0.7, 0.2, 0.1],
    [0.2, 0.6, 0.2],
    [0.1, 0.3, 0.6],
    [0.5, 0.2, 0.3]
], dtype=np.float64)
observed = np.asarray([0, 1, 2, 2, 0]) 
def viterbi(initial_prob, transition, emission_prob, observed, room_ids = ["1", "2", "3", "4"]):
    delta = [initial_prob * emission_prob[:,observed[0]]]
    psi = [np.asarray([0]*3)]
    for i in range(1, len(observed)):
        # psi.append(np.asarray([
        #      np.argmax(delta[i-1]*transition[:,j])
        # for j in range(3)]))
        # delta.append(np.asarray([
        #      emission_prob[j][observed[i]]*np.max(delta[i-1]*transition[:,j])
        # for j in range(3)]))
        psi.append(np.argmax(delta[i-1]*transition.transpose(),axis=1))
        delta.append(emission_prob[:,observed[i]]*np.max(delta[i-1]*transition.transpose(),axis=1))

    for i in range(len(observed)):
        print(f"delta[{i+1}]={delta[i]}")

    max_prob = np.max(delta[len(delta)-1])
    predictions = [np.argmax(delta[len(delta)-1])]
    for i in range(len(observed)-1,0,-1):
        predictions.append(psi[i][predictions[len(predictions)-1]])

    room_predictions = [room_ids[pred] for pred in reversed(predictions)]
    print(f"max probability {max_prob} for sequence:\n{room_predictions}")
```

```python
# a)
viterbi(initial_prob, transition, emission_prob, observed)

# b)
transition2 = np.asarray([
    [0.0,0.5,0.0,0.5],
    [0.5,0.0,0.5,0.0],
    [0.0,0.5,0.0,0.5],
    [0.5,0.0,0.5,0.0],
], dtype=np.float64)
viterbi(initial_prob, transition2, emission_prob, observed)

# desi secventa cea mai probabila de camere este la fel pentru ambele subpuncte,
# probabilitatea comuna a acestei secvente de camere si a observatiilor este putin mai mare (0.0003 -> 0.0008)
```
