Se aruncă două zaruri, iar rezultatele sunt notate cu X şi Y.
1. Calculaţi probabilitatea ca suma să fie cel puţin 10, ştiind că primul zar indică un număr par.
2. Calculaţi probabilitatea ca suma să fie cel puţin 10, ştiind că cel puţin unul dintre zaruri indică un număr par.
3. Calculaţi probabilitatea ca ambele zaruri să indice numere pare, ştiind că suma lor este 8.
4. Sunt evenimentele A = {X + Y ≥ 10} şi B = {X este par} independente?

Extensie Python: Simulaţi 100.000 de aruncări şi estimaţi probabilităţile condiţionate. Comparaţi rezultatele cu val-
orile exacte.



1. Folosing notația de la subpunctul 4:$$P(A|B)=\frac{P(A \cap B)}{P(B)}=\frac{\frac{4}{36}}{\frac{1}{2}}=\frac{2}{9}$$

2. Considerăm evenimentul $C=\{Y \text{este par}\}$:
$$P(A|(B \cup C))=\frac{P(A \cap (B \cup C))}{P(B \cup C)}=\frac{\frac{5}{36}}{\frac{3}{4}}=\frac{5}{27}$$


3. Considerăm evenimentul $D=\{X+Y=8\}$:

$$P(B \cap C | D)=\frac{P(B \cap C \cap D)}{P(D)}=\frac{\frac{3}{36}}{\frac{5}{36}}=\frac{3}{5}$$


4. $P(A)=\frac{6}{36}$, $P(B)=\frac{1}{2}$

$$P(A \cap B)=\frac{4}{36} \neq P(A) \cdot P(B) \Rightarrow \text{$A$ și $B$ nu sunt independente}$$

```python
import numpy as np

def run()->tuple[int]:
    return (np.random.randint(1,7), np.random.randint(1,7))

def runs(iter_cnt: int)->None:
    count_a_and_b = 0
    count_b = 0
    count_a_and_b_or_c = 0
    count_b_or_c = 0
    count_b_and_c_and_d = 0
    count_d = 0
    for _ in range(iter_cnt):
        X, Y = run()
        A = (X+Y>=10)
        B = (X%2==0)
        C = (Y%2==0)
        D = (X+Y==8)
        if A and B: count_a_and_b += 1
        if B: count_b += 1
        if A and (B or C): count_a_and_b_or_c += 1
        if (B or C): count_b_or_c += 1
        if B and C and D: count_b_and_c_and_d += 1
        if D: count_d += 1
    print(f"P(A|B)={count_a_and_b/count_b}")
    print(f"P(A|(B U C))={count_a_and_b_or_c/count_b_or_c}")
    print(f"P(B,C|D))={count_b_and_c_and_d/count_d}")

runs(100000)
```
